```python
import re
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

stop_words = set(stopwords.words("english"))

# Keep negation words because they're important for sentiment
stop_words = stop_words - {"no", "not", "nor"}
```

```python
def preprocess_text(text):

    # 1. Convert to lowercase
    text = text.lower()

    # 2. Keep English letters and spaces
    text = re.sub(r"[^a-zA-Z\s]", " ", text)

    # 3. Remove extra spaces
    text = re.sub(r"\s+", " ", text).strip()

    # 4. Tokenization
    tokens = word_tokenize(text)

    # 5. Remove stopwords
    tokens = [
        word for word in tokens
        if word not in stop_words
    ]

    # 6. Lemmatization
    tokens = [
        lemmatizer.lemmatize(word)
        for word in tokens
    ]

    # 7. Join tokens back into a sentence
    clean_text = " ".join(tokens)

    return clean_text
```

```python
sample = "This product is NOT good!!! The batteries are very poor."

processed = preprocess_text(sample)

print("Original:")
print(sample)

print("\nProcessed:")
print(processed)
```

```python
original = df["full_review"].iloc[0]

processed = preprocess_text(original)

print("ORIGINAL:")
print(original)

print("\nCLEANED:")
print(processed)
```

```python
df["clean_text"] = df["full_review"].apply(preprocess_text)
```
